1.1. Что такое GitHub?
GitHub — это веб-платформа для хостинга IT-проектов и совместной разработки, основанная на системе контроля версий Git. Она предоставляет облачное хранилище репозиториев, инструменты для code review, управления задачами, CI/CD, а также социальные функции (звёзды, форки, обсуждения).

1.2. Как GitHub связан с Git?
Git — это распределённая система контроля версий (работает локально).
GitHub — это сервис, который использует Git как основу, добавляя к нему веб-интерфейс, удалённое хранение репозиториев и функции для командной работы (pull requests, issues, actions). Команды git clone, git push, git pull работают с удалёнными репозиториями на GitHub.

1.3. Чем отличается fork репозитория от его клонирования (clone)?
Fork cоздаёт копию репозитория на самом GitHub (в вашем аккаунте). Clone создаёт локальную копию репозитория на вашем компьютере.

1.4. Зачем нужны и как работают pull requests?
Pull request (PR) — механизм предложения изменений в репозиторий.

1.5. GitHub использует ваш почтовый адрес для привязки ваших Git коммитов к вашей учётной записи?
Да, но не всегда автоматически.
GitHub связывает коммиты с вашей учётной записью, если email в конфигурации Git (user.email) совпадает с одним из подтверждённых email-адресов в настройках GitHub.

3.1 Какие практики принято соблюдать при создании Pull Request, чтобы закрыть автоматический issues?
    Какие практики принято соблюдать при создании commit, чтобы закрыть автоматический issues?

При создании Pull Request нужно указать в описании PR (или в тексте коммита) одно из ключевых слов, за которыми следует номер issue.
Указывать номер issue в описании PR (не только в заголовке):
text
##Описание
Исправлена ошибка валидации формы.
Closes #15
При создании коммита (в сообщении) используются те же ключевые слова — GitHub анализирует сообщения коммитов при их попадании в основную ветку.
Включать ключевое слово и номер issue в тело коммита (лучше, чем в заголовок):
bash
git commit -m "Исправлена ошибка с кэшированием
Реализована очистка кэша после обновления данных.
Fixes #89"

4.Можно ли открыть пул реквест, если вы ничего не вносили в FORK?
Нет, нельзя. Чтобы открыть пул реквест, в вашем форке должны быть изменения (коммиты), которых нет в исходном репозитории. Если вы ничего не меняли, GitHub не покажет кнопку "Create pull request", так как нечего будет сливать.

4.1 Что нужно сделать чтобы открыть пул реквест? (Рисунок 90)
Сначала сделайте Fork нужного репозитория (кнопка в правом верхнем углу)
В своём форке внесите изменения (через редактирование файлов, загрузку или локально через Git)
После этого GitHub покажет баннер с предложением создать пул реквест
Нажмите "Compare & pull request"
Убедитесь, что базовый репозиторий — исходный, а сравниваемая ветка — ваша
Напишите заголовок и описание изменений
Нажмите "Create pull request"

4.2 Что нужно сделать если ваш форк устарел?
Чтобы обновить устаревший форк и потом открыть чистый пул реквест:
Откройте свой форк на GitHub
Нажмите "Sync fork" (кнопка над списком веток)
Нажмите "Update branch"
Альтернатива (через командную строку):
bash
git remote add upstream [URL исходного репозитория]
git fetch upstream
git checkout main
git merge upstream/main
git push origin main

4.3 Что нужно сделать если в пул реквесте имеются конфликты слияния (Рисунок 96)
Откройте пул реквест со страницы Pull requests
Прокрутите вниз до раздела с сообщением о конфликтах
Есть три способа решения:
Через GitHub (для простых конфликтов): нажмите "Resolve conflicts" → отредактируйте файлы в веб-редакторе → нажмите "Mark as resolved"
Локально через командную строку: выполните git pull origin main, устраните конфликты вручную, сделайте git add ., git commit, git push
Через GitHub Desktop: откройте PR, нажмите "Open in Desktop" и разрешите конфликты через интерфейс программы

5.Что нужно сделать для добавления отрывка кода в комментарии к ишьюсу? (Рисунок 104)
Напишите комментарий в ишьюсе или пул реквесте
Чтобы вставить блок кода, используйте тройные обратные кавычки (```) до и после кода
После открывающих кавычек можно указать язык (например, ```python) для подсветки синтаксиса
Также можно использовать одинарные обратные кавычки (`) для встроенных фрагментов кода
Нажмите "Comment"

5.1 На какую клавишу нажать, чтобы выделенный текст был включён как цитата в ваш комментарий? (Рисунок 105)
Нажмите клавишу > (знак "больше") — она автоматически превратит выделенный текст в цитату. Либо нажмите "Quote reply" (кнопка "три точки" рядом с комментарием, который хотите процитировать).

5.2 Как вставить картинку в ишьюс? (Рисунок 108)
Способы вставки изображения в комментарий:
Перетащите файл изображения из папки прямо в поле комментария (самый простой способ)
Скопируйте изображение (Ctrl+C) и вставьте (Ctrl+V) в поле комментария
Нажмите на иконку "Attach files" (скрепка) и выберите файл на компьютере
Если изображение уже в интернете — используйте Markdown-синтаксис: ![Alt текст](URL картинки)

6.Как понять что ваш форк устарел?
Признаки того, что форк устарел:
На странице вашего форка под кнопкой "Sync fork" появляется сообщение "This branch is [N] commits behind [исходный репозиторий]:main"
При попытке создать пул реквест GitHub предупреждает "Can't automatically merge"
Баннер "This branch is out-of-date with the base branch"

6.1 Как обновить форк?
Через интерфейс GitHub:
Перейдите в свой форк
Нажмите "Sync fork" (над списком веток)

Через командную строку (более надёжно):
bash
git remote add upstream https://github.com/оригинал-владелец/репозиторий.git
git fetch upstream
git checkout main

git push origin main

7.Как добавить участников в ваш репозиторий, чтобы команда могла работать над одним репозиторием? (Рисунок 112)
Откройте ваш репозиторий на GitHub
Перейдите в Settings → Collaborators and teams (или просто Collaborators)
Нажмите "Add people" (зелёная кнопка)
Введите имя пользователя, email или имя команды

9.Что такое и зачем нужен файл README
README — это первый файл, который видят посетители репозитория (обычно в формате .md — Markdown). Он нужен для:
Описания проекта: что делает программа/библиотека
Инструкции по установке и запуску
Примеров использования
Информации о лицензии
Ссылок на документацию
GitHub автоматически отображает содержимое README на главной странице репозитория.

9.1 Что такое и зачем нужен файл CONTRIBUTING (Рисунок 122)
CONTRIBUTING (или CONTRIBUTING.md) — это файл с правилами для тех, кто хочет внести свой вклад в проект. Он нужен для:
Объяснения, как правильно оформлять пул реквесты
Описания код-стайла и требований к тестам
Инструкции по запуску проекта в режиме разработки
Этики взаимодействия (Code of Conduct)
GitHub показывает ссылку на этот файл, когда кто-то пытается создать ишьюс или пул реквест.

10.Как изменить основную ветку (Рисунок 123)
Перейдите в Settings вашего репозитория
На боковой панели выберите Branches (или "General" → вкладка "Branches")
В разделе "Default branch" нажмите на иконку карандаша ✏️
Выберите новую ветку из выпадающего списка
Подтвердите действие по инструкциям GitHub (может потребоваться обновить ветки в пул реквестах)

10.1 Как передать проект? какая кнопка? (рисунок 124)
Для передачи проекта другому пользователю:
Перейдите в Settings → General
Прокрутите вниз до раздела "Danger Zone"
Нажмите "Transfer ownership" (или "Transfer repository")
Введите имя нового владельца и название репозитория
Подтвердите действие

10.2 Что такое файл .gitignore?
.gitignore — это специальный текстовый файл, который указывает Git, какие файлы и папки НЕ нужно отслеживать (не добавлять в репозиторий). В него обычно записывают:
Системные файлы (.DS_Store, Thumbs.db)
Папки зависимостей (node_modules/, venv/, vendor/)
Конфигурации с секретами (.env, config.local.json)
Скомпилированные файлы (*.class, *.pyc, *.exe)
Лог-файлы (*.log)
Пример содержимого:
text
#Python
__pycache__/
*.pyc
venv/

#Node.js
node_modules/
.env

#IDE
.vscode/
.idea/
