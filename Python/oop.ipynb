```python
class PersonHello:
    """Класс с методом без self, вызываемым через сам класс."""

    name = "Ivan"

    @staticmethod
    def hello() -> None:
        """Поздороваться."""
        print("Hello")
```

```python
PersonHello.hello()
```

```python
class PersonCreate:
    """Класс, задающий атрибут через отдельный метод create()."""

    def create(self) -> None:
        """Задать имя Ivan."""
        self.name = "Ivan"  # pylint: disable=attribute-defined-outside-init

    def display(self) -> None:
        """Вывести имя."""
        print(self.name)
```

```python
person_create = PersonCreate()
```

```python
person_create.create()
```

```python
person_create.display()
```

```python
class PersonInit:
    """Класс, задающий имя через конструктор."""

    def __init__(self, name: str) -> None:
        """Сохранить имя."""
        self.name = name

    def display(self) -> None:
        """Вывести имя."""
        print(self.name)
```

```python
person_init = PersonInit("Ivan")
person_init.name
```

```python
class PersonStatic:
    """Класс с обычным методом и статическим методом."""

    def hello(self) -> None:
        """Поздороваться."""
        print("Hello")

    @staticmethod
    def goodbye() -> None:
        """Попрощаться."""
        print("Goodbye")
```

```python
person_static = PersonStatic()
```

```python
person_static.hello()
```

```python
person_static.goodbye()
```

```python
person_a = PersonStatic()
person_b = PersonStatic()
```

```python
id(person_a.hello)
```

```python
id(person_b.hello)
```

```python
id(person_a.goodbye)
```

```python
id(person_b.goodbye)
```
