# GitHub

1.1. Что такое GitHub?
GitHub - это веб-платформа для хостинга Git-репозиториев, то есть сервис, где хранятся проекты с полной историей их изменений (коммитов), с возможностью совместной работы над кодом: обсуждения (Issues), предложение изменений (Pull Requests), автоматизация (Actions), управление доступом участников и многое другое поверх обычного git.

1.2. Как GitHub связан с Git?
Git - это сама технология контроля версий (программа, которая работает локально на компьютере и отслеживает изменения в файлах, коммиты, ветки). GitHub - это отдельный сервис, который просто хранит git-репозитории в облаке и добавляет удобный веб-интерфейс поверх git. То есть Git можно использовать вообще без GitHub (например, хранить репозиторий только локально или на своём собственном сервере), а GitHub, в свою очередь, физически работает именно через команды git (clone, push, pull, fetch).

1.3. Чем отличается fork репозитория от его клонирования (clone)?
**Fork**: создаёт твою собственную копию чужого репозитория на самом GitHub (в облаке), с полными правами записи именно в эту копию — нужен, когда у тебя нет прав пушить напрямую в оригинал

**Clone**: скачивает содержимое репозитория (любого — своего, форкнутого, или того, к которому у тебя есть прямой доступ) на твой локальный компьютер, чтобы можно было работать с файлами в IDE

1.4. Зачем нужны и как работают pull requests?
Pull Request - это формальное предложение влить твои изменения (обычно из отдельной ветки или форка) в другую ветку или репозиторий. 

Нужен, чтобы:

- дать возможность review — другие участники (или владелец репозитория) могут посмотреть код, оставить комментарии, до того как изменения попадут в основную ветку
- запустить автоматические проверки (CI — линтеры, тесты) перед мёржем
- вести историю обсуждения конкретного изменения — почему оно было сделано, какие были вопросы

1.5. GitHub использует ваш почтовый адрес для привязки ваших Git коммитов к вашей учётной записи?
Да. Каждый git-коммит содержит в себе имя автора и email, указанные в локальной конфигурации git (git config user.email). GitHub сопоставляет этот email с email, привязанным к аккаунту пользователя — если они совпадают, коммит на GitHub отображается с аватаркой и ссылкой на профиль автора; если email не совпадает ни с одним аккаунтом — коммит покажется "анонимным", без привязки к конкретному пользователю GitHub, даже если по факту его сделал реальный зарегистрированный человек.

1.6 Какая команда генерирует SSH ключ для Доступа по SSH к репозиторию (Рисунок 83)
ssh-keygen -t ed25519 -C "твой_email@example.com"

После генерации публичный ключ (id_ed25519.pub) добавляется в настройках GitHub (Settings - SSH and GPG keys - New SSH key), а приватный ключ остаётся только на компьютере. Это позволяет делать git push/git pull по SSH-протоколу без необходимости каждый раз вводить логин/пароль или токен.

2.9 как открыть запрос слияния, указывающий на другой запрос слияния и зачем это нужно? (Рисунок 117)

Это функциональность связывания PR друг с другом. Например, когда одна фича разбита на несколько последовательных PR, и второй PR технически базируется на ветке первого (ещё не смёрженного). Делается это через изменение base branch при создании PR. Вместо base: main указываешь base: ветка-другого-PR, тем самым явно показывая, что твой PR "стоит поверх" ещё не влитого. Нужно это, чтобы не блокировать параллельную разработку — можно продолжать работу над второй частью фичи, не дожидаясь, пока первую часть смёржат, и review проходит по каждой части отдельно, без смешивания диффов.

# Рабочий процесс с использованием GitHub
3. Напишите 8 пунктов, которые нужно сделать, чтобы внести вклад в чужой проект.

1) Найти проект, в который хочешь внести вклад
2) Сделать Fork репозитория (создать свою копию на GitHub)
3) Clone — склонировать свой форк на компьютер
4) Создать отдельную ветку под конкретную задачу (например, dev или fix-typo)
5) Внести изменения в код и закоммитить их с понятным сообщением (по конвенции коммитов)
6) Запушить изменения в свой форк (git push)
7) Открыть Pull Request из своего форка в оригинальный репозиторий
8) Дождаться review/CI-проверок и, при необходимости, доработать по замечаниям, пока PR не будет принят (смёржен)

3.1.

- Какие практики принято соблюдать при создании Pull Request чтобы закрыть автоматический issues?

1) В описании PR использовать ключевое слово + номер issue: Closes #12, Fixes #12 или Resolves #12
2) Указывать это в описании самого PR, а не только в заголовке — GitHub ищет эти слова именно в теле/описании
3) Один PR может закрывать сразу несколько issues, если перечислить несколько таких строк

- Какие практики принято соблюдать при создании commit чтобы закрыть автоматический issues?

1) То же самое ключевое слово в сообщении коммита: Fixes #12
2) Важно, чтобы такой коммит попал име  нно в основную ветку через мёрдж — тогда issue закроется автоматически в момент мёржа
3) Можно писать это как в заголовке коммита, так и в расширенном описании (git commit -m "заголовок" -m "Closes #12")

3.2 Как отклонить/закрыть пул реквест? (предоставьте скриншот где это в гитхабе)

На странице PR, внизу — кнопка "Close pull request" (красная, рядом с полем для комментария). Это закрывает PR без слияния, то есть отклоняет предложенные изменения, в отличие от "Merge pull request", который бы их принял.

# Создание запроса на слияние

4 Можно ли открыть пул реквест, если вы ничего не вносили в FORK?

Чисто технически, да, можно открыть PR, даже если форк идентичен оригиналу (нет изменений). Но такой PR будет пустым — не покажет никакого diff'а, и GitHub, скорее всего, не даст его создать осмысленно (или создаст, но он будет бессодержательным, "нечего сливать"). По сути PR имеет смысл только тогда, когда есть реальные изменения (коммиты) в ветке форка, которых нет в целевой ветке.

4.1 Что нужно сделать чтобы открыть пул реквест? (Рисунок 90)

1. Внести изменения и закоммитить их в ветке своего форка
2. Запушить эту ветку на GitHub (git push)
3. Зайти на страницу PR (Pull requests, затем нажать New pull request)
4. Выбрать base repository (куда вливать, обычно оригинал) и head repository (откуда, мой форк) + нужные ветки
5. Заполнить заголовок и описание
6. Нажать "Create pull request"

4.2 Что нужно сделать Если ваш Форк устарел?

Нужно синхронизировать его с оригиналом — либо через кнопку "Sync fork" прямо на странице форка на GitHub (самый простой способ), либо вручную через терминал: 

```
git fetch upstream
git merge upstream/main (или git pull upstream main) 
git push origin main
```

4.3 Что нужно сделать если в пул реквесте имеются конфликты слияния (Рисунок 96)

Разрешить их. Либо через встроенный редактор GitHub (кнопка "Resolve conflicts" на странице PR), либо локально через git: подтянуть актуальную целевую ветку, смёржить её в свою рабочую ветку, вручную разрешить конфликты в редакторе (Accept Current/Incoming/Both), закоммитить и запушить снова

# Поддержание GitHub репозитория в актуальном состоянии

6 Как понять что ваш форк устарел?
На странице форка на GitHub, под названием репозитория, обычно отображается надпись вида "This branch is X commits behind [оригинал]:main" — прямое указание, на сколько коммитов твой форк отстаёт от оригинала.


6.1 Как обновить форк?

Кнопка "Sync fork" прямо на странице форка (появляется именно тогда, когда есть отставание) — нажимаешь, и GitHub сам подтягивает и вливает изменения из оригинала в твой форк, без необходимости лезть в терминал.

# Особенные файлы

9 Что такое и зачем нужен файл README

README.md — главный файл описания проекта, который GitHub автоматически показывает на главной странице репозитория. Нужен, чтобы объяснить, что делает проект, как его установить и использовать, какие есть возможности. Это первое, что видит человек, зашедший в репозиторий.

9.1 Что такое и зачем нужен файл CONTRIBUTING (Рисунок 122)

CONTRIBUTING.md — файл с правилами и инструкциями для тех, кто хочет внести вклад в проект: как оформлять issues/PR, каким стандартам кода следовать, как настроить окружение для разработки, какой процесс review принят в проекте. Помогает контрибьюторам не гадать, а следовать чёткому, задокументированному процессу.

# Управление проектом

10 Как изменить основную ветку (Рисунок 123)

Settings репозитория, раздел "General", блок "Default branch". Нажимаем на значок переключения/карандаш рядом с текущей веткой, выбираем новую и подтверждаем смену.

10.1 Как передать проект? какая кнопка? (рисунок 124)

Settings репозитория, в самом низу страницы, раздел "Danger Zone" и нажимаем на кнопку "Transfer ownership".

10.2 Что такое файл .gitignore?

Файл, в котором перечислены пути/паттерны файлов и папок, которые git должен игнорировать, то есть не отслеживать и не включать в коммиты, даже если они физически лежат в папке проекта.
