   8.  Что содержится в папке include в CPython?

   9. Где можно найти исходный код CPython дайте ссылку на репозиторий гитхаб

   10. (опционально) Как работает интерпретатор CPython при выполнении кода?
    1) Исходный код парсится и формируется абстрактное синтаксическое дерево
    2) АСТ компилируется в байт код
    3) Выполняет инструкции байт кода: стек, вызовы, объекты, ветвления
    4) Повторяющиеся модули кешируются, чтобы не делать лишнюю компиляцию
