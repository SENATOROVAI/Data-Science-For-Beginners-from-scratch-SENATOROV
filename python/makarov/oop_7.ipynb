```python
class Animal:

    def __init__(self, weight, lenght):
        self.weight = weight
        self.lenght = lenght

    def eat(self):
        print("Eating")

    def sleep(self):
        print("Sleeping")
```

```python
class Bird(Animal):

    def move(self):

        print("Flying")
```

```python
pigeon = Bird(0.3, 30)
```

```python
pigeon.weight, pigeon.lenght
```

```python
pigeon.eat()
```

```python
pigeon.move()
```

```python
class Bird(Animal):

    def __init__(self, weight, lenght, flying_speed):
        super().__init__(weight, lenght)
        self.flying_speed = flying_speed

    def move(self):
        print("Flying")
```

```python
pigeon = Bird(0.3, 30, 100)
```

```python
pigeon.weight, pigeon.lenght, pigeon.flying_speed
```

```python
pigeon.sleep()
```

```python
pigeon.move()
```

```python
class CatClass:

    def __init__(self, name, color):
        self.name = name
        self._type_ = "кот"
        self.color = color

    def info(self):
        print(f"Меня зовут {self.name}, я {self._type_}, цвет моей шерсти {self.color}")

    def sound(self):
        print("Я умею мяукать")
```

```python
class DogClass:

    def __init__(self, name, color):
        self.name = name
        self._type_ = "пес"
        self.color = color

    def info(self):
        print(f"Меня зовут {self.name}, я {self._type_}, цвет моей шерсти {self.color}")

    def sound(self):
        print("Я умею лаять")
```

```python
cat = CatClass("Бегемот", "черный")
dog = DogClass("Барбос", "серый")
```

```python
for animal in (cat, dog):
    animal.info()
    animal.sound()
    print()
```

```python
patients = [
    {"name": "Николай", "height": 178},
    {"name": "Иван", "height": 182},
    {"name": "Алексей", "height": 190},
]
```

```python
total, count = 0, 0

for patient in patients:

    total += patient["height"]

    count += 1

total / count
```

```python
class DataClass:

    def __init__(self, data):
        self.data = data

    def count_average(self, metric):

        self.metric = metric

        self.__total = 0
        self.__count = 0

        for item in self.data:

            self.__total += item[self.metric]

            self.__count += 1

        return self.__total / self.__count
```

```python
data_object = DataClass(patients)

data_object.count_average("height")
```

```python
heights = list(map(lambda x: x["height"], patients))

heights
```

```python
sum(heights) / len(heights)
```
