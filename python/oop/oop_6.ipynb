```python
"""Static methods and decorator."""
```

```python
class Person:
    def hello(self):
        print("Hello")

    @staticmethod
    def goodbye():
        print("Goodbye")


p = Person()
p.goodbye()
```

```python
a = Person()
```

```python
b = Person()
```

```python
a.hello()
```

```python
a.goodbye()
```

```python
id(a.hello)
```

```python
id(b.hello)
```

```python
id(a.goodbye)
```

```python
id(b.goodbye)
```
