23.3) Различие между стэшем и коммитом.  
  Когда лучше сохранить изменения в стэше, а когда коммитить.

  Stash - это временно сохраненные в буфер обмена или "карман" результаты работы с кодом при принятии Pull (изменений рабочего репозитория) или при временной остановке работы. 
  Коммит - это полное сохранение собственных изменений рабочего кода, перед окончанием работы или готовых изменений к отправке на проверку или перед публикацией. 
  Сохранять изменения в Stash лучше при принятии новой версии кода, чтобы увидеть изменения и принять решение дальнейшей консолидации или выбора версии репозитория, с которой будем работать дальше, а коммитить лучше, когда нужно зафиксировать результаты работы. (12:14 или https://t.me/c/1937296927/3602/17092)

23.6) Практические примеры использования стэша.
