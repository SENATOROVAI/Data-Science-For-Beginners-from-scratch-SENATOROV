23.3) Различие между стэшем и коммитом.  
коммит - это просто загрузка изменений в репозиторий на сервере. а стэш - это буфер, где хранятся конфликтующие изменения до решения, что с ними делать
  Когда лучше сохранить изменения в стэше, а когда коммитить.
  если работа с кодом еще не завершена, ее рано коммитить, то пока ее лучше просто сохранить
