23.6) Практические примеры использования стэша.  
  Краткие сценарии, где стэш помогает. Стэш помогает разрешить конфликт слияния: локальные изменения заносятся в стэш, тем временем скачиваются изменения из интернета, позже решается, какие изменения оставить, а какие - удалить. (сценарий из видео)
  Еще примеры из интернета: 
  1) Срочное переключение на другую задачу (например, нужно срочно починить баг, а коммитить сломанный код не хочется, тогда можно убрать его в стэш)
  2) Перенос незакоммиченных правок на другую ветку (стэш с перепутанной ветки -> перейти на другую ветку -> вставить изменения из стэша, где изначально предпологалось, будет писаться ваш код)
  3) Тестирование альтернативного решения (Эксперименты) (вернуться с помощью стэша к исходной версии кода, если эксперимент (например, улучшение алгоритма) провалился)
